```python
import pandas as pd
from sklearn.model_selection import train_test_split
from sklearn.linear_model import LogisticRegression
from sklearn.tree import DecisionTreeClassifier
from sklearn.metrics import accuracy_score, confusion_matrix, recall_score,precision_score,f1_score,classification_report
from sklearn.preprocessing import LabelEncoder
```

```python
df=pd.read_csv("iris.csv")
df.head()
```

```python
encoder = LabelEncoder()
df['Species'] = encoder.fit_transform(df['Species'])
```

```python
df.head()
```

```python
x=df.drop('Species',axis=1)
y=df['Species']
```

```python
x_train,x_test,y_train,y_test=train_test_split(x,y,test_size=0.3,random_state=25)
```

```python
lr=LogisticRegression()
lr.fit(x_train,y_train)
y_pred=lr.predict(x_test)
print("Accuracy of Logistic Regression",accuracy_score(y_test,y_pred))

print("Logistic Regression Confusion Matrix\n")
pd.DataFrame(confusion_matrix(y_test,y_pred),columns=list(range(0,3)))
```

```python
dt=DecisionTreeClassifier()
dt.fit(x_train,y_train)
y_pred1=dt.predict(x_test)
print(accuracy_score(y_test,y_pred1))

print("Decision Tree Confusion Matrix\n")
pd.DataFrame(confusion_matrix(y_test,y_pred1),columns=list(range(0,3)))
```

```python
result = pd.DataFrame()
result['Actual Label'] = y_test
result['Logistic Regression Prediction'] = y_pred
result['Decision Tree Prediction'] = y_pred1
result.sample(10)
```

```python
precision_score(y_test,y_pred,average=None)
```

```python
recall_score(y_test,y_pred,average=None)
```

```python
precision_score(y_test,y_pred1,average='weighted')
```

```python
recall_score(y_test,y_pred1,average='weighted')
```

```python
f1_score(y_test,y_pred1,average='weighted')
```

```python
print(classification_report(y_test,y_pred1))
```
